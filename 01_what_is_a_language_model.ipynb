# What is a language model?

> Scotland’s nature is among our greatest assets.
> 
> Nature gives us food and drinking water, energy and timber.

[_What we do | NatureScot_](https://www.nature.scot/about-naturescot/our-work/what-we-do)

## Break it into a chain of words

> Scotland’s nature is among our greatest assets.

```mermaid
flowchart LR

A(scotland’s) --> B(nature)
B --> C(is)
C --> D(among)
D --> E(our)
E --> F(greatest)
F --> G(assets)
```

## Repeated words lead to forks in the road and loops

> Nature gives us food and drinking water, energy and timber.

```mermaid
flowchart LR

H(nature) --> I(gives)
I --> J(us)
J --> K(food)
K --> L(and)
L --> M(drinking)
M --> N(water)
N --> O(energy)
O --> L
L -->P(timber)

```
